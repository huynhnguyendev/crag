```python
from typing import List, TypedDict
import time
import re

from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores import FAISS
from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings,
)
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate

from langgraph.graph import StateGraph, START, END
from pydantic import BaseModel
from dotenv import load_dotenv

load_dotenv()
```

## **Xử lý phần tải file, chia chunks, lưu vào vector stores (FAISS)**

```python
# 1) Load the PDF document
docs = (
    PyPDFLoader("../documents/AI_Engineer_Roadmap_2026.pdf").load()
)

print("Pages in document:", len(docs))
```

```python
# 2) Chunk
chunks = RecursiveCharacterTextSplitter(
    chunk_size=900, 
    chunk_overlap=150
).split_documents(docs)

print("Chunks:", len(chunks))
```

```python
# 3) Embedding
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
)
```

```python
# 4) Create FAISS vector store
vector_store = FAISS.from_documents(
    chunks,
    embeddings
)
```

```python
# 5) Create retriever
retriever = vector_store.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0,
)
```

```python
class State(TypedDict):
    question: str
    docs: List[Document]

    strips: List[str]            # tách docs ra
    kept_strips: List[str]       # lọc và giữ lại
    refined_context: str         # ghép lại

    answer: str
```

```python
def retrieve(state: State) -> State:
    query = state["question"]
    output = retriever.invoke(query)

    return {
        "docs": output
    }
```

```python
# -----------------------------
# Sentence-level DECOMPOSER
# -----------------------------
def decompose_to_sentences(text: str) -> List[str]:
    # Chuẩn hóa khoảng trắng:
    # - Thay nhiều khoảng trắng liên tiếp bằng 1 khoảng trắng
    # - Xóa khoảng trắng ở đầu và cuối text
    text = re.sub(r"\s+", " ", text).strip()

    # Tách text thành từng câu dựa trên các dấu:
    # ".", "!" hoặc "?"
    #
    # Ví dụ:
    # "RAG là gì? RAG kết hợp retrieval và generation."
    #
    # sẽ trở thành:
    # [
    #     "RAG là gì?",
    #     "RAG kết hợp retrieval và generation."
    # ]
    sentences = re.split(r"(?<=[.!?])\s+", text)

    # Tạo danh sách để lưu những câu hợp lệ
    result = []

    # Duyệt qua từng câu
    for sentence in sentences:

        # Xóa khoảng trắng thừa ở đầu và cuối câu
        sentence = sentence.strip()

        # Chỉ giữ lại những câu có độ dài lớn hơn 20 ký tự
        if len(sentence) > 20:
            result.append(sentence)

    # Trả về danh sách các câu sau khi đã xử lý
    return result


# -----------------------------
# FILTER (LLM judge)
# -----------------------------

# Schema quy định format output mà LLM phải trả về
#
# Ví dụ:
# {
#     "keep": true
# }
#
# hoặc:
# {
#     "keep": false
# }
class KeepOrDrop(BaseModel):
    keep: bool


# Prompt dùng để yêu cầu LLM đánh giá
# một sentence có liên quan trực tiếp đến question hay không
filter_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a strict relevance filter.\n"
            "Return keep=true only if the sentence directly helps answer the question.\n"
            "Use ONLY the sentence. Output JSON only.",
        ),
        (
            "human",
            "Question: {question}\n\n"
            "Sentence:\n{sentence}",
        ),
    ]
)


# Tạo LangChain pipeline:
#
# filter_prompt
#      ↓
# Gemini LLM
#      ↓
# KeepOrDrop structured output
#
# with_structured_output() giúp LLM trả kết quả
# theo schema KeepOrDrop thay vì text tự do.
filter_chain = filter_prompt | llm.with_structured_output(KeepOrDrop)


# -----------------------------
# REFINING
# Decompose -> Filter -> Recompose
# -----------------------------
def refine(state: State) -> State:

    # Lấy câu hỏi của user từ State
    q = state["question"]

    # Lấy nội dung text từ tất cả documents được retrieve
    #
    # Document có dạng:
    # Document(
    #     page_content="...",
    #     metadata={...}
    # )
    #
    # Ở đây chỉ lấy phần page_content.
    context = ""

    for doc in state["docs"]:
        context += doc.page_content
        context += "\n\n"

    # Xóa khoảng trắng thừa ở đầu và cuối context
    context = context.strip()

    # ---------------------------------
    # 1. DECOMPOSITION
    # ---------------------------------
    # Chia context lớn thành từng sentence nhỏ
    strips = decompose_to_sentences(context)

    # ---------------------------------
    # 2. FILTER
    # ---------------------------------
    # Danh sách chứa những sentence
    # được LLM đánh giá là có liên quan
    kept = []

    # Duyệt qua từng sentence
    for sentence in strips:

        # Gửi question + sentence cho Gemini
        result = filter_chain.invoke(
            {
                "question": q,
                "sentence": sentence,
            }
        )

        # Nếu Gemini đánh giá sentence có liên quan
        if result.keep:
            kept.append(sentence)

    # ---------------------------------
    # 3. RECOMPOSE
    # ---------------------------------
    # Ghép những sentence được giữ lại
    # thành một context hoàn chỉnh
    refined_context = ""

    for sentence in kept:
        refined_context += sentence
        refined_context += "\n"

    # Xóa newline/khoảng trắng thừa ở cuối
    refined_context = refined_context.strip()

    # ---------------------------------
    # RETURN
    # ---------------------------------
    # Lưu các kết quả trung gian vào State
    return {
        "strips": strips,
        "kept_strips": kept,
        "refined_context": refined_context,
    }
```

```python
answer_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a helpful ML tutor. Answer ONLY using the provided refined bullets.\n"
            "If the bullets are empty or insufficient, say: 'I don't know based on the provided books.'",
        ),
        ("human", "Question: {question}\n\nRefined context:\n{refined_context}"),
    ]
)

def generate(state: State) -> State:

    # Lấy câu hỏi từ State
    question = state["question"]

    # Lấy context đã được lọc và tinh chỉnh
    refined_context = state["refined_context"]

    # Chuẩn bị dữ liệu đầu vào cho prompt
    prompt_input = {
        "question": question,
        "refined_context": refined_context,
    }

    # Tạo chain: Prompt → LLM
    chain = answer_prompt | llm

    # Gửi prompt_input vào chain
    response = chain.invoke(prompt_input)

    # Lấy nội dung câu trả lời từ LLM
    answer = response.content

    # Gemini có thể trả content dưới dạng list[dict]
    # Lấy phần text ra thành string
    if isinstance(answer, list):
        answer = answer[0]["text"]

    # Lưu câu trả lời vào State
    return {
        "answer": answer
    }
```

```python
g = StateGraph(State)

g.add_node("retrieve", retrieve)
g.add_node("refine", refine)
g.add_node("generate", generate)

g.add_edge(START, "retrieve")
g.add_edge("retrieve", "refine")
g.add_edge("refine", "generate")
g.add_edge("generate", END)

app = g.compile()
app
```

```python
res = app.invoke({
    "question": "Hãy tóm tắt các chương cần học để trở thành một AI Engineer trong năm 2026, theo roadmap trong cuốn sách này.",
    "docs": [],
    "strips": [],
    "kept_strips": [],
    "refined_context": "",
    "answer": ""
})
print(res["answer"])
```
